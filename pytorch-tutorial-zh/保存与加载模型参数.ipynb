保存与加载模型
=======================

在这一节中，将研究如何保存模型状态，加载和运行模型预测


```python
import torch
import torchvision.models as models
```


保存和加载模型权重
================================

PyTorch模型在内部状态字典中保存了学习的参数，字段叫`state_dict`。可以通过`torch.save`方法持久化：

```python
model = models.vgg16(weights='IMAGENET1K_V1')
torch.save(model.state_dict(), 'model_weights.pth')
```

为了加载模型权重，需要先创建一个相同的模型。然后通过`load_state_dict()`方法加载参数

在以下代码中，我们设置`weights_only=True`来限制反序列化过程中执行的函数，只加载必要的参数。通过`weights_only=True`来加载权重是最佳实践。

```python
model = models.vgg16() # we do not specify ``weights``, i.e. create untrained model
model.load_state_dict(torch.load('model_weights.pth', weights_only=True))
model.eval()
```

保存和加载模型形状
=====================================

当家在模型权重时我们需要先实例化模型类，因为类定义了模型的网络结构。我们也需要保存模型的网络结构，对此可以传入`model`而非`model.state_dict()`到保存函数


```python
torch.save(model, 'model.pth')
```

可以通过如下所示的方式加载模型

像链接里所说的[Saving and loading torch.nn.Modules](https://pytorch.org/docs/main/notes/serialization.html#saving-and-loading-torch-nn-modules),保存`state_dict`是最佳实践，但是下main我们用`weights_only=False`，因为这也包括了加载模型，这是`torch.save`的遗留用例



```python
model = torch.load('model.pth', weights_only=False)
```

<div style="background-color: #54c7ec; color: #fff; font-weight: 700; padding-left: 10px; padding-top: 5px; padding-bottom: 5px"><strong>NOTE:</strong></div>

<div style="background-color: #f3f4f7; padding-left: 10px; padding-top: 10px; padding-bottom: 10px; padding-right: 10px">

<p>这个办法序列化模型时用了 <a href="https://docs.python.org/3/library/pickle.html">的序列化方法</a> 模块,因此，在加载模型时它依赖真实的类定义</p>
